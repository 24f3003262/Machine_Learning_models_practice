Stemming means reducing a word to its word stem that affixes to prefixes or suffixes or to roots of the words called lemma.

```python
words = [
    "cacti", "focuses", "went", "is", "rocking", 
    "phenomena", "better", "studying", "corpora", 
    "stolen", "mice", "probabilities"
]
```

Porter Stemmer

```python
from nltk.stem import PorterStemmer
```

```python
stemming=PorterStemmer()
```

```python
for word in words:
    print(word+"----->"+stemming.stem(word))
```

Stemming works for a lot of words but also does not for a lot of words. Changes the meaning in a bad way very heavily

```python
from nltk.stem import RegexpStemmer
```

```python
reg_stemmer=RegexpStemmer('ing$|s$|e$|able$',min=4)
```

```python
reg_stemmer.stem('eating')
```

```python
reg_stemmer.stem('ingeating')
```

Snowball Stemmer

```python
from nltk.stem import SnowballStemmer
```

```python
snowball=SnowballStemmer('english')
```

```python
for word in words:
    print(word+"---->"+snowball.stem(word))
```

porter vs snowball

```python
stemming.stem('fairly'),stemming.stem('sportingly')
```

```python
snowball.stem('fairly'),snowball.stem('sportingly')
```

```python
snowball.stem('goes')
```
