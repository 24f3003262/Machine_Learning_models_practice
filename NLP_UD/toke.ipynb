```python
corpus=''' Humanity is utterly fucked. 
I am the one who shall solve it all.
I am stronger faster better.
-Homelander
'''
```

```python
print(corpus)
```

```python
from nltk.tokenize import sent_tokenize
```

```python
documents=sent_tokenize(corpus)
```

```python
type(documents)
```

```python
for sentence in documents:
    print(sentence)
```

### TOKENIZATIOn
Paragraph to words
sentence to words

```python
from nltk.tokenize import word_tokenize
```

```python
words=word_tokenize(corpus)
```

```python
for word in words:
    print(word)
```

```python
from nltk.tokenize import wordpunct_tokenize
```

```python
wordpunct_tokenize(corpus) ## same as word tokenize but punctuation is also splitted
```

```python
from nltk.tokenize import TreebankWordTokenizer
```

```python
tolenizer=TreebankWordTokenizer()
```

```python
tolenizer.tokenize(corpus)
```
